```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print("Dataset loaded successfully!")
```

```python
print("\nFirst 10 rows:")
display(df.head(10))

print("\nLast 10 rows:")
display(df.tail(10))
```

```python
print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
print("\nDataset Information:")
df.info()

```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["Unit_Price"] = pd.to_numeric(
    df["Unit_Price"],
    errors="coerce"
)

df["Total_Sales"] = pd.to_numeric(
    df["Total_Sales"],
    errors="coerce"
)

print("\nData types converted successfully!")
```

```python
total_sales = df["Total_Sales"].sum()

total_orders = df["Order_ID"].nunique()

total_quantity = df["Quantity"].sum()

average_order_value = total_sales / total_orders


print("\n========== KEY SALES METRICS ==========")

print("Total Sales:",
      round(total_sales, 2))

print("Total Orders:",
      total_orders)

print("Total Quantity Sold:",
      total_quantity)

print("Average Order Value:",
      round(average_order_value, 2))

```

```python
print("\n========== SALES SUMMARY ==========")

display(
    df["Total_Sales"].describe()
)
```

```python
print("\n========== TOP PRODUCTS ==========")

top_products = (
    df.groupby("Product")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

display(
    top_products.reset_index(
        name="Total_Sales"
    )
)
```

```python
print("\n========== TOP 10 PRODUCTS ==========")

top_10_products = top_products.head(10)

display(
    top_10_products.reset_index(
        name="Total_Sales"
    )
)
```

```python
print("\n========== SALES BY CATEGORY ==========")

category_analysis = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values(
          "Total_Sales",
          ascending=False
      )
)

display(category_analysis)
```

```python
print("\n========== CATEGORY SALES PERCENTAGE ==========")

category_analysis["Sales_Percentage"] = (
    category_analysis["Total_Sales"]
    / total_sales
) * 100

display(category_analysis)
```

```python
print("\n========== SALES BY REGION ==========")

region_analysis = (
    df.groupby("Region")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values(
          "Total_Sales",
          ascending=False
      )
)

display(region_analysis)
```

```python
print("\n========== REGION SALES PERCENTAGE ==========")

region_analysis["Sales_Percentage"] = (
    region_analysis["Total_Sales"]
    / total_sales
) * 100

display(region_analysis)
```

```python
print("\n========== REGION + CATEGORY ANALYSIS ==========")

region_category = (
    df.groupby(
        ["Region", "Category"]
    )["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

display(region_category)
```

```python
print("\n========== REGION-CATEGORY PIVOT TABLE ==========")

region_category_pivot = pd.pivot_table(
    df,
    values="Total_Sales",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)

display(region_category_pivot)
```

```python
print("\n========== MONTHLY SALES ==========")

df["Year"] = df["Order_Date"].dt.year

df["Month"] = df["Order_Date"].dt.month

df["Month_Name"] = df["Order_Date"].dt.strftime("%B")


monthly_sales = (
    df.groupby(
        ["Year", "Month", "Month_Name"]
    )["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values(
        ["Year", "Month"]
    )
)

display(monthly_sales)
```

```python
print("\n========== MONTHLY ORDERS ==========")

monthly_orders = (
    df.groupby(
        ["Year", "Month", "Month_Name"]
    )["Order_ID"]
    .nunique()
    .reset_index(
        name="Number_of_Orders"
    )
)

display(monthly_orders)
```

```python
print("\n========== PAYMENT METHOD ANALYSIS ==========")

payment_analysis = (
    df.groupby("Payment_Method")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values(
          "Total_Sales",
          ascending=False
      )
)

display(payment_analysis)
```

```python
top_region = region_analysis[
    "Total_Sales"
].idxmax()

top_region_sales = region_analysis[
    "Total_Sales"
].max()

print("\nTop Region:", top_region)

print(
    "Top Region Sales:",
    round(top_region_sales, 2)
)
```

```python
top_category = category_analysis[
    "Total_Sales"
].idxmax()

top_category_sales = category_analysis[
    "Total_Sales"
].max()

print("\nTop Category:", top_category)

print(
    "Top Category Sales:",
    round(top_category_sales, 2)
)
```

```python
top_product = top_products.idxmax()

top_product_sales = top_products.max()

print("\nTop Product:", top_product)

print(
    "Top Product Sales:",
    round(top_product_sales, 2)
)

```

```python
highest_order = df.loc[
    df["Total_Sales"].idxmax()
]

print("\n========== HIGHEST VALUE ORDER ==========")

display(
    highest_order.to_frame().T
)

```

```python
lowest_order = df.loc[
    df["Total_Sales"].idxmin()
]

print("\n========== LOWEST VALUE ORDER ==========")

display(
    lowest_order.to_frame().T
)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    category_analysis.index,
    category_analysis["Total_Sales"]
)

plt.title("Total Sales by Category")

plt.xlabel("Category")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    region_analysis.index,
    region_analysis["Total_Sales"]
)

plt.title("Total Sales by Region")

plt.xlabel("Region")

plt.ylabel("Total Sales")

plt.tight_layout()

plt.show()

```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_10_products.index[::-1],
    top_10_products.values[::-1]
)

plt.title("Top 10 Products by Sales")

plt.xlabel("Total Sales")

plt.ylabel("Product")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales["Month_Name"],
    monthly_sales["Total_Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    payment_analysis.index,
    payment_analysis["Total_Sales"]
)

plt.title("Sales by Payment Method")

plt.xlabel("Payment Method")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.imshow(
    region_category_pivot,
    aspect="auto"
)

plt.colorbar(
    label="Total Sales"
)

plt.xticks(
    range(len(region_category_pivot.columns)),
    region_category_pivot.columns,
    rotation=45
)

plt.yticks(
    range(len(region_category_pivot.index)),
    region_category_pivot.index
)

plt.title("Sales by Region and Category")

plt.xlabel("Category")

plt.ylabel("Region")

plt.tight_layout()

plt.show()
```

```python
print("\n========== FINAL SUMMARY ==========")

summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Orders",
        "Total Quantity Sold",
        "Average Order Value",
        "Top Product",
        "Top Category",
        "Top Region"
    ],

    "Value": [
        round(total_sales, 2),
        total_orders,
        total_quantity,
        round(average_order_value, 2),
        top_product,
        top_category,
        top_region
    ]
})

display(summary)
```

```python
print("\n========== KEY FINDINGS ==========")

best_month = monthly_sales.loc[
    monthly_sales["Total_Sales"].idxmax(),
    "Month_Name"
]

best_month_sales = monthly_sales[
    "Total_Sales"
].max()

top_payment = payment_analysis[
    "Total_Sales"
].idxmax()


print(
    "1. Total sales generated:",
    round(total_sales, 2)
)

print(
    "2. Total number of orders:",
    total_orders
)

print(
    "3. Total quantity sold:",
    total_quantity
)

print(
    "4. Average order value:",
    round(average_order_value, 2)
)

print(
    "5. Top-selling product:",
    top_product
)

print(
    "6. Highest-performing category:",
    top_category
)

print(
    "7. Highest-performing region:",
    top_region
)

print(
    "8. Highest-sales month:",
    best_month
)

print(
    "9. Most-used/highest-sales payment method:",
    top_payment
)

```

```python
output_file = "CodeOrbit_Task2_Sales_Analysis.xlsx"

with pd.ExcelWriter(output_file) as writer:

    summary.to_excel(
        writer,
        sheet_name="Key Metrics",
        index=False
    )

    top_products.reset_index(
        name="Total_Sales"
    ).to_excel(
        writer,
        sheet_name="Top Products",
        index=False
    )

    category_analysis.reset_index().to_excel(
        writer,
        sheet_name="Category Analysis",
        index=False
    )

    region_analysis.reset_index().to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    region_category.to_excel(
        writer,
        sheet_name="Region Category",
        index=False
    )

    monthly_sales.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    payment_analysis.reset_index().to_excel(
        writer,
        sheet_name="Payment Analysis",
        index=False
    )

print("\nExcel analysis file created successfully!")
```
